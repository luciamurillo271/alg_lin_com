def rota(theta) : #Mision cumplida
Recibe un angulo theta y retorna una matriz de 2x2 que rota un vector dado en un angulo theta

```python
def rota(theta):
    a = np.cos(theta)
    b = np.sin(theta)
    A = np.array([[a,-b],[b,a]])
    return A

print(rota(np.pi/2))
```

def escala(s): #Mision cumplida
Recibe una tira de numeros s y retorna una matriz cuadrada de n x n, donde n es el tamano de s. La matriz escala la componente i de un vector de Rn en un factor s[ i ]

```python
def escala(s):
    A = np.zeros((len(s),len(s)))
    for i in range(len(s)):
        A[i][i] = s[i]
    return A
```

def rota_y_escala (theta,s): #Mision cumplida
Recibe un angulo theta y una tira de numeros s, y retorna una matriz de 2x2 que rota el vector en un angulo theta y luego lo escala en un factor s

```python
def rota_y_escala(theta,s):
    T2 = escala(s)
    T1 = rota(theta)
    return T2 @ T1
```

def afin(theta,s,b) : #Mision cumplida
Recibe un angulo theta , una tira de numeros s (en R2) , y un vector b en R2. Retorna una matriz de 3x3 que rota el vector en un angulo theta, luego lo escala en un factor s y por ultimo lo mueve en un valor fijo b

```python
def afin(theta,s,b):
    T2 = escala(s)
    T1 = rota(theta)
    A = T2@T1
    B = np.zeros((3,3))
    for i in range(2):
        B[i][:2] = A[i]
        B[i][2] = b[i]
    B[2][2] = 1
    return B
```

def trans afin(v,theta ,s ,b) : #Mision cumplida
Recibe un vector v (en R2) , un angulo theta, una tira de numeros s (en R2), y un vector b en R2. Retorna el vector w resultante de aplicar la transformacion afin a v

```python
def trans_afin(v,theta,s,b):
    A = afin(theta,s,b)
    v1 = np.array([v[0],v[1],1])
    w = A@v1
    return w[:2]
```
